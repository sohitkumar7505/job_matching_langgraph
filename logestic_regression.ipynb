```python
import math

X = [1, 2, 3, 4, 5]
Y = [0, 0, 0, 1, 1]


b0 = 0
b1 = 0
lr = 0.1
epochs = 1000

n = len(X)


def sigmoid(z):
    return 1 / (1 + math.exp(-z))


for _ in range(epochs):
    sum_error_b0 = 0
    sum_error_b1 = 0

    for i in range(n):
        x = X[i]
        y = Y[i]

        z = b0 + b1 * x
        p = sigmoid(z)

        error = p - y

        sum_error_b0 += error
        sum_error_b1 += error * x


    b0 -= lr * (sum_error_b0 / n)
    b1 -= lr * (sum_error_b1 / n)

print("b0 (bias):", b0)
print("b1 (weight):", b1)
```
